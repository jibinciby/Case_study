```python
df = pd.read_csv("Sample - Superstore.csv", encoding="cp1252")
```

```python
#1.What is the size and structure of the dataset?
print("Size of dataset: ",df.size)
print("Structure of dataset: ",df.shape)
```

```python
#2.Are there missing values or duplicate records?
print("Missing values:",df.isnull().sum().sum())
print("Duplicate records:",df.duplicated().sum())
```

```python
#3.What are the total sales, total profit, and average order-line sales?
print("Total sales:",df["Sales"].sum())
print("Total profit:",df["Profit"].sum())
print("Average order line sales:",df["Sales"].mean())
```

```python
#4.Which product categories generate the most sales?
most_sales=df.groupby("Category")["Sales"].sum().sort_values(ascending=False)
print("most_sales:\n",most_sales)
```

```python
#5.Which sub-categories generate the highest and lowest profit?
sub=df.groupby("Sub-Category")["Profit"].sum().sort_values(ascending=False)
print("sub-category with highest profit:",sub.idxmax())
print("sub-category with lowest profit:",sub.idxmin())
```

```python
#6.Which region has the highest sales?
region=df.groupby("Region")["Sales"].sum().sort_values(ascending=False)
print("region with the highest sales:",region.idxmax())
```

```python
#7.Which customer segment contributes the most sales?
customer=df.groupby("Segment")["Sales"].sum().sort_values(ascending=False)
print("customer who contributes the most sales is:",customer.idxmax())
```

```python
#8.How do discounts relate to profit?
print("Discount-profit relation:\n",df.groupby("Discount")["Profit"].mean())
```

```python
#9.How do sales change over time?
df['Order Date']=pd.to_datetime(df['Order Date'])
sales=df.groupby("Order Date")["Sales"].sum()
print(sales)
```
