```python
import pandas as pd
```

```python
#1.
scores=pd.Series([78,85,92,67,74,88,95,61], index=['Anu','Binu','Cathy','Deepa','Ebin','Fathima','Gopal','Hari'])
print(scores)
```

```python
#2
dept=dict({'Anu':'CS','Binu':'IT','Cathy':'CS','Deepa':'BCA','Ebin':'IT','Fathima':'CS','Gopal':'CS','Hari':'BCA'})
print(dept)
```

```python
#3
scores.name='Test Score'
print(scores)
```

```python
#4
print('values :',scores.values)
print('index :',scores.index)
print('data type :',scores.dtype)
print('name :',scores.name)
```

```python
#5
print(f'cathys score: {scores.loc['Cathy']}')
```

```python
#6
print(f'score of first student: {scores.iloc[0]}')
```

```python
#7
print(f'scores from Binu to Ebin:\n {scores.loc['Binu':'Ebin']}')
print(f'scores of first three students:\n {scores.iloc[0:3]}')
```

```python
#8
print(f'scores of Anu, Cathy and Ebin:\n {scores.loc['Anu':'Ebin':2]}')
```

```python
#9
print(f'Highest score is: {scores.max()}')
print(f'Lowest score is: {scores.min()}')
print(f'Mean score is: {round(scores.mean(),2)}')
print(f'number of students: {len(scores)}')
```

```python
#10
print(f'scores greater than 80:\n {scores[scores>=80]}')
```

```python
#11
s=scores.sort_values(ascending=False)
print(f'top 3 scorers:\n {s[0:3]}')
```

```python
#12
print(f'student with highest score: {s.idxmax()}')
```

```python
#13
def grade(scores):
  if scores>=90:
    return 'Distinction'
  elif scores>=75:
    return 'First class'
  elif scores>=60:
    return 'Second Class'
  else:
    return 'Fail'
res=scores.apply(grade)
print(res)
```

```python
#14
print(res.value_counts())
print(res.value_counts(normalize=True))
```

```python
#15
a=res.unique()
print(f'unique grades: {a}')
print(f'number of unique grades: {len(a)}')
```

```python
#16
bonus_scores=lambda scores:scores+5
print(f'after adding bonus 5:\n {scores.apply(bonus_scores)}')
print(f'old scores:\n {scores}')
```
