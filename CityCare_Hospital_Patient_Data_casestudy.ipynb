```python
df=pd.read_csv("Hospital_Patients.csv")
```

```python
#PART A
#1.Load the file and take a first look at the data: first and last rows, a random sample, shape, columns and data types. List every problem you can spot (wrong data types, missing values, anything odd). Age is shown as 'object' - explain why.
print("first row:\n",df.head(1))
print("last row:\n",df.tail(1))
print("random row:\n",df.sample(1))
print("shape:\n",df.shape)
print("columns:\n",df.columns)
print("data types:\n",df.dtypes)
```

```python
#Age has the wrong data type
#Missing values
#Incorrect AdmitDate data type
```

```python
#PART B
#2.Produce summary statistics for the numeric columns and for all columns. Report the mean, median, mode, minimum, maximum, standard deviation, count and sum of StayDays. What is the total number of patient-days?
print("Summary statistics for numeric columns:\n",df.describe())
print("\nSummary statistics for all columns:\n",df.describe(include='all'))
print("\nMean:", df["StayDays"].mean())
print("\nMedian:", df["StayDays"].median())
print("\nMode:", df["StayDays"].mode())
print("\nMinimum:", df["StayDays"].min())
print("\nMaximum:", df["StayDays"].max())
print("\nStandard Deviation:", df["StayDays"].std())
print("\nCount:", df["StayDays"].count())
print("\nSum:", df["StayDays"].sum())

print("\nTotal Patient-Days:", df["StayDays"].sum())
```

```python
#3.Compare the mean and median of BillAmount. Which one better represents a typical bill, and why?
print("Mean Bill Amount:", df["BillAmount"].mean())
print("\nMedian Bill Amount:", df["BillAmount"].median())

#median represents typicall bill because, the mean is significantly higher due to an unusually large bill.
```

```python
#PART C
#4.How many patients are in each Department (count and percentage)? What percentage of patients have insurance? How many unique values does Gender have?
dept_count = df["Department"].value_counts()
print("no of Patients in Each Department:\n",dept_count)
dept_percentage = df["Department"].value_counts(normalize=True) * 100
print("\nno of Patients in Each Department in percentage:\n",dept_percentage)
insurance_percentage = (df["Insurance"] == "Yes").mean() * 100
print("\nInsurance percentage: ",insurance_percentage)
print("\nNumber of Unique Gender Values:", df["Gender"].nunique())
```

```python
#PART D
#5.Create a missing-value report showing the count and percentage for each column. Decide how to treat BillAmount and SatisfactionScore, and fill both with the median. Justify why the mean or zero is not suitable.
missing_report = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": (df.isnull().sum() / len(df)) * 100
})
print(missing_report)
df["BillAmount"] = df["BillAmount"].fillna(df["BillAmount"].median())
df["SatisfactionScore"] = df["SatisfactionScore"].fillna(df["SatisfactionScore"].median())
print("\nMissing Values After Filling:\n",df[["BillAmount", "SatisfactionScore"]].isnull().sum())
```

```python
#PART E
#6.Detect, display and remove the duplicate patient records. Then check that PatientID is unique.
duplicates = df[df.duplicated(keep=False)]
df = df.drop_duplicates()
print("Duplicate Patient Records:\n", duplicates,
      "\n\nCleaned Dataset:\n", df)
print(f"\n\nIs PatientID unique? {df['PatientID'].is_unique}")
```

```python
#PART F
#7.Convert Age to a number. Some ages are the text 'unknown' - decide how to handle them. After conversion, check the missing values again and explain why new missing values appeared. Fill them using the median age and convert Age to int.
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
print("missing values:\n",df.isnull().sum())
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Age"] = df["Age"].astype(int)
print("Age:\n",df["Age"])
```

```python
#8.Convert AdmitDate to datetime so that invalid dates become NaT. How many records have invalid dates? Also convert Department, Gender and Insurance to the 'string' type
df["AdmitDate"] = pd.to_datetime(df["AdmitDate"], errors="coerce")
print("Invalid dates:", df["AdmitDate"].isnull().sum())
df["Department"] = df["Department"].astype("string")
df["Gender"] = df["Gender"].astype(str)
df["Insurance"] = df["Insurance"].astype(str)
print("datatypes:\n",df.dtypes)
```

```python
#PART G
#9.Which patients stayed more than 10 days? Which patients are unhappy (SatisfactionScore below 5)? Which uninsured patients have a bill above Rs. 50,000?
print("Patients who stayed more than 10 days:\n",df[df["StayDays"] > 10])
print("\nPatients who are unhappy:\n",df[df["SatisfactionScore"] < 5])
print("\nUninsured patients with a bill above Rs. 50,000:\n",df[(df["Insurance"] == "No") & (df["BillAmount"] > 50000)])
```

```python
#10.Show the 5 most recently admitted patients and the 5 patients with the highest BillAmount. Does the highest bill look suspicious?
df["AdmitDate"] = pd.to_datetime(df["AdmitDate"], errors="coerce")
print("Five Most Recently Admitted Patients:\n",df.sort_values("AdmitDate", ascending=False).head(5))
print("\nFive Patients with the Highest Bill Amount:\n",df.sort_values("BillAmount", ascending=False).head(5))
```

```python
#PART H
#11.For each Department, find the average StayDays and average BillAmount. For each Department, also give the count, mean, median, minimum and maximum of BillAmount. For Insurance Yes/No, find the average SatisfactionScore. Which department is the most expensive?
print("Average StayDays and BillAmount:\n",df.groupby('Department')[['StayDays', 'BillAmount']].mean())
print("BillAmount statistics:\n",df.groupby('Department')['BillAmount'].agg(['count', 'mean', 'median', 'min', 'max']))
print("Average SatisfactionScore based on Insurance",df.groupby('Insurance')['SatisfactionScore'].mean())
print("Most expensive department:",df.groupby('Department')['BillAmount'].mean().idxmax())
```

```python
#PART I
#12.Find the correlation between StayDays and BillAmount, between StayDays and SatisfactionScore, and between Age and StayDays. Explain what each value means for the hospital.
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')
result1 = df['StayDays'].corr(df['BillAmount'])
print("Correlation between StayDays and BillAmount:", result1)
result2 = df['StayDays'].corr(df['SatisfactionScore'])
print("Correlation between StayDays and SatisfactionScore:", result2)
result3 = df['Age'].corr(df['StayDays'])
print("Correlation between Age and StayDays:", result3)
```

```python
#PART J
#13.Detect outliers in BillAmount using the IQR method. Show the limits and the outlier rows. Then compute the correlation between StayDays and BillAmount again after removing the outlier. What changed? Which version would you trust?
Q1=df["BillAmount"].quantile(0.25)
Q3=df["BillAmount"].quantile(0.75)

IQR=Q3-Q1

lower=Q1-1.5*IQR
upper=Q3+1.5*IQR

outliers=df[
    (df["BillAmount"]<lower) |
    (df["BillAmount"]>upper)
    ]
print(outliers)
print("lower limit:",lower)
print("upper limit:",upper)
corr_before = df["StayDays"].corr(df["BillAmount"])
print("correlation before:",corr_before)
df_clean = df.drop(outliers.index)
corr_after = df_clean["StayDays"].corr(df_clean["BillAmount"])
print("correlation after:",corr_after)
```

```python
#PART K
#14.Create Cost_Per_Day as BillAmount divided by StayDays. Which Department has the highest average Cost_Per_Day?
df["Cost_Per_Day"] = df["BillAmount"] / df["StayDays"]
print("Cost Per Day for Each Patient:\n",df[["PatientID", "Department", "Cost_Per_Day"]])
avg_cost = df.groupby("Department")["Cost_Per_Day"].mean()
print("Average Cost Per Day for Each Department:\n",avg_cost)
highest = avg_cost.idxmax()
print("Department with highest cost per day:", highest)
```

```python
#15.Using lambda functions, create Stay_Type: 'Short' for 3 days or fewer, 'Medium' for 4-7 days, and 'Long' for more than 7 days. Create Age_Group: 'Child' (below 18), 'Adult' (18-59) and 'Senior' (60 and above). Count each category.
stay = lambda x: "Short" if x <= 3 else ("Medium" if x <= 7 else "Long")
df["Stay_Type"] = df["StayDays"].apply(stay)
print("Stay Type:\n",df[["StayDays", "Stay_Type"]])
age = lambda x: "Child" if x < 18 else ("Adult" if x < 60 else "Senior")
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
df["Age_Group"] = df["Age"].apply(age)
print(df[["Age", "Age_Group"]])
print("Count of Each Stay Type:",df["Stay_Type"].value_counts())
print("Count of Each Age Group:",df["Age_Group"].value_counts())
```

```python
#16.Find the average SatisfactionScore for each Stay_Type and each Age_Group. What pattern do you see?
print("Average Satisfaction Score for Each Stay Type:\n",df.groupby("Stay_Type")["SatisfactionScore"].mean())
print("Average Satisfaction Score for Each Age Group:\n",df.groupby("Age_Group")["SatisfactionScore"].mean())
```

```python
#PART L
#17.Export df_clean to cleaned_hospital.csv without the index, then reload the file to verify its shape and data types.
df_clean.to_csv("cleaned_hospital.csv", index=False)
df_new = pd.read_csv("cleaned_hospital.csv")
print("shape of new dataframe:",df_new.shape)
print("datatypes of new dataframe:\n",df_new.dtypes)
```

```python
#FINAL QUESTIONS
#1.Which department needs attention - the one with the highest cost per day, or the one with the lowest satisfaction?
cost = df.groupby("Department")["Cost_Per_Day"].mean()
print("Average Cost Per Day by Department:\n",cost.sort_values(ascending=False))
satisfaction = df.groupby("Department")["SatisfactionScore"].mean()
print("Average Satisfaction Score by Department:\n",satisfaction.sort_values())
```
