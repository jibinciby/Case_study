```python
import numpy as np
```

```python
employees=np.array(["Arun","Bhavana","Charan","Divya","Eshan","Fathima"])
performance=np.array([
    [78,82,85,88],
    [65,70,68,74],
    [92,89,95,91],
    [55,62,60,67],
    [81,79,84,87],
    [72,75,78,80]
])
quarters=np.array(["Q1","Q2","Q3","Q4"])
```

```python
#1
#Calculate the total performance score for every employee.
total_performance_score=performance.sum(axis=1)
print(f' total performance score for every employee are: {total_performance_score}')
```

```python
#2
#Calculate the average quarterly performance for every employee.
average_quarterly_performance=performance.mean(axis=1)
print(f'average quarterly performance for every employee are: {average_quarterly_performance}')
```

```python
#3
#Calculate the average performance for each quarter.
average_performance_quaterly=performance.mean(axis=0)
print(f'average performance for each quarter are: {average_performance_quaterly}')
```

```python
#4
#Identify the employee with the highest total performance score.
maximum=max(total_performance_score)
highest=employees[total_performance_score==maximum]
print(f'employee with the highest total performance score is: {highest}')
```

```python
#5
#Rank all employees based on their total performance score.
print(employees)
print(total_performance_score)
a=np.argsort(total_performance_score)
print(f'employees ranked based on their total performance score: {employees[a]}')
```

```python
#6
#Calculate the quarter-over-quarter percentage growth for every employee from Q1 to Q4.
q1q2 =(performance[:, 1]-performance[:, 0])/performance[:, 0]*100
q2q3=(performance[:,2]-performance[:,1])/performance[:,1]*100
q3q4=(performance[:,3]-performance[:,2])/performance[:,2]*100
print(f'quarter-over-quarter percentage growth for every employee from Q1 to Q2 is:  {q1q2}')
print(f'quarter-over-quarter percentage growth for every employee from Q2 to Q3 is:  {q2q3}')
print(f'quarter-over-quarter percentage growth for every employee from Q3 to Q4 is:  {q3q4}')
```

```python
#7
#Identify employees whose average performance is greater than or equal to 80.
print(f'employees whose average performance is greater than or equal to 80 are: {employees[average_quarterly_performance>=80]}')
```

```python
#8
#Count how many employees have an average performance below 70.
below70=average_quarterly_performance[average_quarterly_performance<70]
print(f'number of employees having an average performance below 70 are: {len(below70)}')
```

```python
#9
#Find the highest performance score recorded in the entire dataset.
highest_performance_score=np.max(performance)
print(f'highest performance score recorded in the entire dataset is: {highest_performance_score}')
```
