```python
#21.Find the correlation of Experience with Salary, Attendance with PerformanceScore, and Age with Experience. Interpret each value (strong / weak, positive / negative).
corr1 = df['Experience'].corr(df['Salary'])
corr2 = df['Attendance'].corr(df['PerformanceScore'])
corr3 = df['Age'].corr(df['Experience'])
print("Correlation b/w Experience and Salary:", corr1)
print("Correlation b/w Attendance and PerformanceScore:", corr2)
print("Correlation b/w Age and Experience:", corr3)
```

```python
#22.Use the IQR method to find the salary outliers. Show the lower and upper limits and the outlier rows. Do you think the outlier is a real salary or a typing error? How can you decide?
Q1=df["Salary"].quantile(0.25)
Q3=df["Salary"].quantile(0.75)

IQR=Q3-Q1

lower=Q1-1.5*IQR
upper=Q3+1.5*IQR

outliers=df[
    (df["Salary"]<lower) |
    (df["Salary"]>upper)
    ]
print(outliers)
print()

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower)
print("Upper Limit:", upper)
```

```python
#23.Remove the outlier and store the result in df_clean. Compute the Experience-Salary correlation again. How much did one extreme value change the result?
corr_before = df['Experience'].corr(df['Salary'])
df_clean = df[
    (df['Salary'] >= lower) &
    (df['Salary'] <= upper)
].copy()
corr_after = df_clean['Experience'].corr(df_clean['Salary'])
print("Correlation before removing outliers:", corr_before)
print("Correlation after removing outliers:", corr_after)
change = corr_after - corr_before
print("Change in correlation:", change)
```

```python
#24.Create a column Overall_Score as the row-wise mean of PerformanceScore and Attendance, and a column Annual_Bonus as 10% of Salary.
df['Overall_Score'] = df[['PerformanceScore', 'Attendance']].mean(axis=1)
df['Annual_Bonus'] = df['Salary'] * 0.10
print(df[['Name', 'PerformanceScore', 'Attendance','Overall_Score', 'Salary', 'Annual_Bonus']])
```

```python
#25.Using a lambda function, create a column Grade: 'Excellent' if PerformanceScore >= 85, 'Good' if PerformanceScore >= 70, otherwise 'Needs Improvement'. Show the count of employees in each grade and the average Salary for each grade.
df['Grade'] = df['PerformanceScore'].apply(
    lambda x: 'Excellent' if x >= 85 else 'Good' if x >= 70 else 'Needs Improvement'
)
print(df[['Name', 'PerformanceScore', 'Grade']])
```
