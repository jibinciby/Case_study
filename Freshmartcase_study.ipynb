```python
#13.Treat the missing values in Discount and Rating using methods you consider appropriate. Save the changes in df, confirm that no missing values remain, and justify your choice of method for each column.
print("missing values in Discount:",df["Discount"].isnull().sum())
print("missing values in Rating:",df["Rating"].isnull().sum())
```

```python
df["Discount"]=df["Discount"].fillna(df["Discount"].median())
df["Rating"]=df["Rating"].fillna(df["Rating"].median())

print("missing values in Discount after changes:",df["Discount"].isnull().sum())
print("missing values in Rating after changes:",df["Rating"].isnull().sum())
```

```python
#26.Find the correlation between Quantity and Discount, and between UnitPrice and Rating. State which relationship is stronger and give a possible reason.
co1=df['Quantity'].corr(df['Discount'])
print("correlation between Quantity and Discount:",co1)
co2=df['UnitPrice'].corr(df['Rating'])
print("correlation between Unitprice and Rating:",co2)
if co1>co2:
  print("Correlation between Quantity and Discount is stronger.")
else:
  print("correlation between Unitprice and Rating is stronger.")
```

```python
#27.Use the Interquartile Range (IQR) method to identify outliers in Quantity. Print Q1, Q3, the lower and upper limits, and the outlier rows.
Q1=df["Quantity"].quantile(0.25)
Q3=df["Quantity"].quantile(0.75)

IQR=Q3-Q1

lower=Q1-1.5*IQR
upper=Q3+1.5*IQR

outliers=df[
    (df["Quantity"]<lower) |
    (df["Quantity"]>upper)
    ]
print(outliers)
print()

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower)
print("Upper Limit:", upper)
```

```python
#28.Examine the outlier rows and decide whether they are genuine or data-entry errors, giving your reasoning. Create a DataFrame named df_clean that excludes the rows you judge to be errors, and compare the mean of Quantity before and after.
print(outliers)

print()
df_clean = df.drop(outliers.index)
mean_before = df['Quantity'].mean()
mean_after = df_clean['Quantity'].mean()
print("Mean Quantity Before Cleaning:", mean_before)
print("Mean Quantity After Cleaning:", mean_after)
```

```python
#29.Use df_clean from this point onwards. Create a column Gross (Quantity x UnitPrice) and a column Revenue, which is Gross after applying the Discount percentage.
df_clean['Gross'] = df_clean['Quantity'] * df_clean['UnitPrice']
df_clean['Revenue'] = df_clean['Gross'] - (df_clean['Gross'] * df_clean['Discount'] / 100)
print(df_clean[['Quantity', 'UnitPrice', 'Discount', 'Gross', 'Revenue']])
```

```python
#30.Create a column Order_Size classifying each order as 'Large' if Revenue is at least 10000, 'Medium' if Revenue is at least 2000 but below 10000, and 'Small' otherwise. Count the orders in each size class.
import numpy as np
df_clean['Order_Size'] = np.where(df_clean['Revenue'] >= 10000, 'Large',np.where(df_clean['Revenue'] >= 2000, 'Medium', 'Small'))
print(df_clean['Order_Size'].value_counts())
```
