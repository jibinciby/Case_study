```python
#A1
#Create a labelled one-dimensional data structure from the enrolment list and give it the name "Course Enrolled". Store it in a variable named enrolments.
enrolments=pd.Series(enrol_list)
enrolments.name='Course Enrolled'
```

```python
#B2
#Display each course's share of total enrolments as a percentage, rounded to 1 decimal place. Store the result in a variable named share.
share=round(enrolments.value_counts(normalize=True),1)
print('share is:\n',share)
```

```python
#C1
#Create a labelled data structure from the study-mode list and give it the name "Study Mode". Store it in a variable named mode.
mode=pd.Series(mode_list)
mode.name='Study Mode'
print(mode)
```

```python
#C2
#Display the number of learners and the percentage share for each study mode.
a=mode.count()
print('number of learners:',a)
b=mode.value_counts(normalize=True)*100
print('percentage share for each study mode:\n',b)
```

```python
#C3
#Identify the most preferred study mode, and compute the combined percentage share of learners who study Online or Hybrid by selecting these two categories together.
c=b.idxmax()
print('most preferred study mode is:',c)
both=b.loc[['Online','Hybrid']]
total=both.sum()
print('combined percentage share of learners who study Online or Hybrid',total)
```

PART D

```python
#D1
#Create a data structure named ratings from the rating list. Display the number of ratings, the average rating (rounded to 2 decimal places), the highest rating and the lowest rating.
ratings=pd.Series(rating_list)
print('number of ratings:',len(ratings))
print('average rating:',round(ratings.mean(),2))
print('highest rating:',ratings.max())
print('lowest rating:',ratings.min())
```

```python
#D2
#Display how many learners gave each rating score.
print(ratings.value_counts())
```

```python
#D3
#Write a function named sentiment(r) that returns "Positive" for a rating of 4 or 5, "Neutral" for a rating of 3, and "Negative" for a rating of 1 or 2. Use it to convert every rating into its sentiment label.
def sentiment(ratings):
  if ratings==4 or ratings==5:
    return "Positive"
  elif ratings==3:
    return "Neutral"
  elif ratings==1 or ratings==2:
    return "Negative"
sentiments=ratings.apply(sentiment)
print(sentiments)
```

```python
#D4
#Display the percentage of learners in each sentiment category, rounded to 1 decimal place.
percent=round(sentiments.value_counts(normalize=True)*100,1)
print(percent)
```
