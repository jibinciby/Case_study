```python
#A Dataframe is a two dimensional labelled table-exactly like a spreadsheet or SQL Table.It has rows(Identified by Index) and coloums(Identified by coloumn labels). Every coloumn of a Dataframe is a Series.
```

```python
#creating Dataframes

data={
    "Name":["Amal","Binu","Cathy","Anna","Tom","Ben"],
    "Dept":["English","CS","Maths","Physics","CS","Maths"],
    "Age":[19,20,22,19,20,22],
    "Marks":[85,72,91,68,88,76]
}
df=pd.DataFrame(data)
print(df)
```

```python
df1=pd.DataFrame(data,index=["s1","s2","s3","s4","s5","s6"])
print(df1)
```

```python
#Exercises
products={
    "Product":["Pen","Notebook","Bag","Bottle","Marker"],
    "Price":[10,45,650,120,30],
    "Stock":[200,45,55,23,12]
}
```

```python
#1.Create a Dataframe pdf from the dictionary products using the labels p1 to p5 as the row index and print it.
pdf=pd.DataFrame(products,index=['p1','p2','p3','p4','p5'])
print(pdf)
```

```python
#2.For pdf display (a) its shape, (b) the columns names and (c) the datatypes of every coloumn.
print("shape\n",pdf.shape)
print("columns\n",pdf.columns)
print("dtype\n",pdf.dtypes)
```

```python
#3.Display the first 3 rows and last 2 rows of pdf.
print("1st 3 rows:\n", pdf.head(3))
print("last 2 rows:\n",pdf.tail(2))
```

```python
#5.Extract the price coloumn of pdf as a series.Display its type, its mean, and the maximum value.Then display the product and stock columns together as a DataFrame.
print(pdf.columns)

a=pdf['Price']
print('price column:\n',a)
print('datatype:',a.dtypes)
print('mean:',a.mean())
print('maximum value:',a.max())

print('product and stock columns:\n',pdf[['Product','Stock']])
```
